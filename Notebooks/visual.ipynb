```python
#importing basics libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()
%matplotlib inline
```

```python
df=pd.read_csv('final_dataset.csv')
df.head()
```

## Price Distribution


```python
sns.histplot(df['Price'], kde=True, color='red')
plt.show()
```

```python
sns.boxplot(df['Price'])
plt.show()
```

## Airline vs Price

```python
sns.boxplot(x='Airline', y='Price', data=df)
plt.xticks(rotation=90)
plt.show()
```

## Source and Destination vs Price

```python
sns.boxplot(x='Source', y='Price', data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
sns.boxplot(x='Destination', y='Price', data=df)
plt.xticks(rotation=90)
plt.show()
```

## Date and Month vs Price

```python
sns.lineplot(x='Date', y='Price', data=df)
plt.show()
```

```python
sns.lineplot(x='Month', y='Price', data=df)
plt.show()
```

## Count of Airlines

```python
plt.figure(figsize=(12, 6))

ax = sns.countplot(x='Airline', data=df, color='orangered')

# Numbers on top of bars
for container in ax.containers:
    ax.bar_label(container, fontsize=10, fontweight='bold', padding=3)

plt.xticks(rotation=90)
plt.title('Number of Flights by Airline', fontsize=16, fontweight='bold')
plt.xlabel('Airline')
plt.ylabel('Count')

plt.show()
```

## Count for Source and Destination

```python
ax = sns.countplot(x='Source', data=df, hue='Source', palette='rocket', legend=False)

for container in ax.containers:
    ax.bar_label(container, fontsize=10, fontweight='bold', padding=3)

plt.show()
```

```python
ax = sns.countplot(x='Destination', data=df, hue='Destination', palette='mako', legend=False)

for container in ax.containers:
    ax.bar_label(container, fontsize=10, fontweight='bold', padding=3)

plt.show()
```
